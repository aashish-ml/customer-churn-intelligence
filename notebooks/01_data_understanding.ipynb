```python
# Check for blank or whitespace-only values in TotalCharges

total_charges_blank = (
    df["TotalCharges"]
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

print(f"Blank/whitespace TotalCharges values: {total_charges_blank}")
```

```python
# Display customers with blank TotalCharges

blank_total_charges = df[
    df["TotalCharges"].astype(str).str.strip().eq("")
]

blank_total_charges[
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

```python
# Check duplicate records

duplicate_rows = df.duplicated().sum()
duplicate_customer_ids = df["customerID"].duplicated().sum()

print(f"Duplicate rows       : {duplicate_rows}")
print(f"Duplicate customerID : {duplicate_customer_ids}")
```

```python
# Check unique values in categorical columns

for column in df.select_dtypes(include="object").columns:
    print(f"\n{column}")
    print(df[column].unique())
```

```python
# Check which TotalCharges values cannot be interpreted as numbers

total_charges_numeric = pd.to_numeric(
    df["TotalCharges"].astype(str).str.strip(),
    errors="coerce"
)

invalid_total_charges = total_charges_numeric.isna().sum()

print(f"TotalCharges values that cannot be converted to numeric: {invalid_total_charges}")
```

```python
# Investigate customers with zero tenure

zero_tenure = df[df["tenure"] == 0]

print(f"Customers with zero tenure: {len(zero_tenure)}")

zero_tenure[
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

```python
# Basic numerical range validation

print("Tenure range:")
print(f"Min: {df['tenure'].min()}")
print(f"Max: {df['tenure'].max()}")

print("\nMonthlyCharges range:")
print(f"Min: {df['MonthlyCharges'].min()}")
print(f"Max: {df['MonthlyCharges'].max()}")
```

```python
categorical_features = [
    column
    for column in df.select_dtypes(include="object").columns
    if column not in ["customerID", "TotalCharges", "Churn"]
]

for column in categorical_features:
    print(f"\n{column}:")
    print(sorted(df[column].dropna().unique().tolist()))
```

```python
print("Unique Churn values:")
print(df["Churn"].unique())

print("\nUnexpected Churn values:")

valid_churn_values = {"Yes", "No"}

unexpected_churn = df[
    ~df["Churn"].isin(valid_churn_values)
]

print(f"Count: {len(unexpected_churn)}")
```
