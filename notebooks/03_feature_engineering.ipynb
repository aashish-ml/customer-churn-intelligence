```python
# Feature Engineering - Setup

import pandas as pd
import numpy as np

# Load raw dataset
df = pd.read_csv("../data/raw/Telco-Customer-Churn.csv")

# Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# Check dataset
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum().sort_values(ascending=False).head(10))
```

```python
# Handle missing values in TotalCharges

print("Missing TotalCharges before:", df["TotalCharges"].isnull().sum())

# Fill missing TotalCharges using median
df["TotalCharges"] = df["TotalCharges"].fillna(
    df["TotalCharges"].median()
)

print("Missing TotalCharges after:", df["TotalCharges"].isnull().sum())
```

```python
print("CustomerID present:", "CustomerID" in df.columns)
print("Dataset Shape:", df.shape)
print("Columns:")
print(df.columns.tolist())
```

```python
# Remove CustomerID column
df = df.drop(columns=["customerID"])

print("customerID removed successfully.")
print("New Dataset Shape:", df.shape)
print("CustomerID present:", "customerID" in df.columns)
```

```python
# Convert Yes/No columns to 1/0

binary_columns = [
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling",
    "Churn"
]

for col in binary_columns:
    df[col] = df[col].map({"Yes": 1, "No": 0})

print("Binary columns converted successfully.")
print(df[binary_columns].head())
```

```python
# Clean restart: reload original dataset

df = pd.read_csv("../data/raw/Telco-Customer-Churn.csv")

print("Original Dataset Shape:", df.shape)

# 1. Remove CustomerID
df = df.drop(columns=["customerID"])

# 2. Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# 3. Fill missing TotalCharges with median
df["TotalCharges"] = df["TotalCharges"].fillna(
    df["TotalCharges"].median()
)

# 4. Convert gender to 0/1
df["gender"] = df["gender"].map({
    "Male": 1,
    "Female": 0
})

# 5. Convert binary Yes/No columns to 0/1
binary_columns = [
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling",
    "Churn"
]

for col in binary_columns:
    df[col] = df[col].map({
        "Yes": 1,
        "No": 0
    })

# 6. One-Hot Encode categorical columns
categorical_columns = [
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaymentMethod"
]

df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)

# 7. Final validation
print("\nFeature engineering completed successfully.")
print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
print("Remaining Object Columns:",
      df.select_dtypes(include="object").columns.tolist())
```

```python
# Remove duplicate rows

print("Duplicate rows before:", df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows after:", df.duplicated().sum())
print("New Dataset Shape:", df.shape)
```
